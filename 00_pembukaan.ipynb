## 1. Tujuan Notebook Ini
- Menyiapkan Python dengan Miniconda di Windows, macOS, atau Linux.
- Menyambungkan environment Python ke VS Code agar file `.ipynb` bisa langsung dijalankan.
- Memberi jalur hemat resource untuk pengguna laptop spesifikasi rendah (laptop kentang).
- Menjelaskan bedanya file `.ipynb` dan `.py` secara singkat.


## 2. Kenapa Pakai Miniconda?
Miniconda ringan karena hanya memasang komponen inti. Cocok untuk pemula dan aman untuk laptop yang RAM/penyimpanannya terbatas.

Prinsipnya: satu proyek satu environment, supaya dependensi rapi dan tidak saling bentrok.


## 3. Instalasi Miniconda (Lintas OS)
1. Buka situs resmi: `https://www.anaconda.com/download/success` lalu pilih **Miniconda** sesuai OS kamu.
2. Install dengan opsi default.
3. Setelah selesai, buka terminal baru lalu cek:
   - `conda --version`

Jika versi tampil, instalasi berhasil.


### Catatan per OS
- Windows: buka **Anaconda Prompt** atau PowerShell.
- macOS: buka Terminal bawaan.
- Linux: buka terminal distro kamu.

Jika `conda` belum terbaca, jalankan `conda init`, tutup terminal, lalu buka lagi.


## 4. Buat Environment Khusus Buku Ini
Jalankan perintah berikut di terminal:


```python
conda create -n pybasic python=3.11 -y
conda activate pybasic
python --version
```

Untuk menjalankan notebook di VS Code, pasang paket minimum ini:


```python
python -m pip install --upgrade pip
pip install ipykernel jupyter
```

## 5. Hubungkan ke VS Code
1. Install VS Code.
2. Install extension: **Python** dan **Jupyter** (dari Microsoft).
3. Buka folder proyek ini di VS Code.
4. Buka file `.ipynb`, klik pemilih kernel (kanan atas), pilih environment `pybasic`.
5. Jalankan sel dengan tombol `Run` atau `Shift+Enter`.


## 6. Jalur Laptop Kentang (Ringan)
Kalau laptop terasa berat, pakai tips ini:
- Tutup aplikasi berat saat belajar (browser tab banyak, meeting app, dll).
- Jalankan notebook satu per satu, jangan banyak notebook aktif sekaligus.
- Pakai paket minimal (`ipykernel`, `jupyter`) dan hindari install library besar dulu.
- Jika VS Code tetap berat, kamu bisa jalankan skrip `.py` dari terminal dulu.


## 7. Bedanya `.ipynb` vs `.py`
- `.ipynb` (Jupyter Notebook): cocok untuk belajar interaktif per sel, ada teks + kode + output dalam satu file.
- `.py` (Python Script): cocok untuk program final/produksi, versi lebih rapi untuk project jangka panjang.

Rekomendasi belajar: pahami konsep di `.ipynb`, lalu biasakan menulis ulang versi `.py` supaya makin kuat.


```python
print('Setup selesai. Selamat belajar Python basic!')
```
