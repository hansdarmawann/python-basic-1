# Bab 6: String


## Tujuan Belajar
- Memahami string sebagai urutan karakter.
- Mengakses karakter dengan indeks dan slicing.
- Menggunakan method string penting.
- Memformat teks dengan f-string.


## 1) Dasar String


```python
teks = "Belajar Python"
print(teks)
print("Panjang:", len(teks))
print("Karakter pertama:", teks[0])
```

## 2) Slicing


```python
teks = "Belajar Python"
print(teks[0:7])   # Belajar
print(teks[8:14])  # Python
print(teks[-6:])   # Python
```

## 3) String Tidak Bisa Diubah Langsung


```python
nama = "andi"
# nama[0] = "A"  # Ini error
nama_baru = "A" + nama[1:]
print(nama_baru)
```

## 4) Method String


```python
kalimat = "  saya suka python  "
print(kalimat.strip())
print(kalimat.upper())
print(kalimat.replace("python", "data"))
print("python" in kalimat)
```

## 5) f-string


```python
nama = "Rani"
umur = 22
print(f"Nama saya {nama}, umur saya {umur} tahun.")
```

## Ringkasan
