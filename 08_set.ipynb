## Tujuan Belajar
- Memahami apa itu `set` dan kapan dipakai.
- Menambah, menghapus, dan mengecek anggota set.
- Menggunakan operasi himpunan seperti union, intersection, difference.


Set itu cocok dipakai ketika kita butuh kumpulan data unik, tanpa duplikasi. Jadi kalau ada data yang sama masuk berkali-kali, set hanya menyimpan satu. Ini berguna banget untuk membersihkan data. Di kode ini, kita lihat langsung perbedaan hasilnya.


```python
# Membuat set dari data yang punya duplikasi
angka = [1, 2, 2, 3, 3, 3, 4]
unik = set(angka)
print("List awal:", angka)
print("Set unik:", unik)
```

Set tidak punya urutan tetap seperti list, jadi akses dengan indeks tidak bisa. Tapi untuk cek keberadaan data, set sangat cepat. Ini sering dipakai saat validasi data atau filter nilai unik. Coba perhatikan cara `in` dipakai di sini.


```python
# Mengecek apakah item ada di dalam set
buah = {"apel", "jeruk", "mangga"}
print("apel" in buah)
print("pisang" in buah)
```

Kita juga bisa menambah dan menghapus item dari set. Pakai `add()` untuk tambah satu data, dan `remove()` atau `discard()` untuk hapus. Bedanya, `remove()` error kalau item tidak ada, sedangkan `discard()` aman. Pilih sesuai kebutuhanmu.


```python
# Menambah dan menghapus data pada set
buah = {"apel", "jeruk"}
buah.add("mangga")
buah.discard("pisang")
print(buah)
```

Operasi himpunan adalah kekuatan utama set. `union` menggabungkan semua item unik, `intersection` mencari item yang sama, dan `difference` mencari item yang hanya ada di satu set. Ini sangat membantu saat membandingkan dua kelompok data. Contohnya bisa kamu pakai untuk membandingkan daftar peserta, produk, atau tag.


```python
# Operasi dasar himpunan
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference a-b:", a - b)
```

Set juga bisa dipakai untuk menghapus duplikasi teks dengan cepat. Di contoh ini, kita pecah kalimat menjadi kata-kata, lalu ubah ke set. Hasilnya hanya kata unik yang tersisa. Teknik ini sering dipakai di preprocessing data teks.


```python
# Mengambil kata unik dari kalimat
kalimat = "python data python coding data"
kata_unik = set(kalimat.split())
print(kata_unik)
```

## Latihan
- Buat set dari daftar nama yang berulang, lalu tampilkan jumlah nama unik.
- Bandingkan dua set hobi temanmu dengan `intersection`.
- Coba bedakan hasil `remove()` dan `discard()` saat data tidak ada.


## Ringkasan
`Set` ideal untuk data unik dan operasi himpunan. Jika fokusmu adalah kecepatan cek anggota dan pembersihan duplikasi, set adalah pilihan yang tepat.
