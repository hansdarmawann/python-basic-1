## 8.2. Dasar Struktur: Cara Mendefinisikan Set
Set dipakai saat kita butuh kumpulan nilai unik. Kamu bisa membuat set langsung dengan `{}` berisi elemen, atau dari `set()` terhadap list/string. Untuk set kosong, wajib pakai `set()` karena `{}` artinya dictionary kosong. Ini detail kecil tapi sering bikin salah paham di awal.


```python
# Cara mendefinisikan set
s1 = {1, 2, 3}
s2 = set([1, 2, 2, 3])
s3 = set("banana")
s4 = set()  # set kosong

print(s1)
print(s2)
print(s3)
print(type(s4))
```

## 8.3. Sifat Set (unik, tidak indexable, mutable)
Set hanya menyimpan item unik, jadi duplikasi otomatis dibuang. Set tidak indexable karena tidak punya urutan tetap seperti list/string. Set bersifat mutable, jadi item bisa ditambah atau dihapus. Karena karakteristik ini, set sangat cocok untuk deduplikasi data.


```python
# Sifat set
angka = {1, 2, 2, 3}
print("Unik:", angka)

angka.add(4)
angka.discard(2)
print("Mutable:", angka)

# print(angka[0])  # error: set tidak indexable
```

## 8.4. Cara Akses Set
Karena set tidak indexable, akses elemen biasanya dilakukan lewat `for` atau pengecekan `in`. Kalau kamu butuh akses berbasis posisi, ubah dulu set ke list. Tapi ingat, urutan set tidak dijamin stabil. Jadi gunakan set untuk keunikan data, bukan untuk urutan.


```python
# Akses set lewat iterasi dan membership test
buah = {"apel", "jeruk", "mangga"}

print("apel" in buah)
for item in buah:
    print(item)

as_list = list(buah)
print("Set jadi list:", as_list)
```

## 8.5. Method Set Penting + Penjelasan
Method set yang penting:
- `add(x)` menambah satu item.
- `update(iterable)` menambah banyak item sekaligus.
- `remove(x)` menghapus item (error kalau item tidak ada).
- `discard(x)` menghapus item (aman kalau item tidak ada).
- `pop()` mengambil item acak dari set.
- `clear()` mengosongkan set.


```python
# Method set penting
huruf = {"a", "b"}
huruf.add("c")
huruf.update(["d", "e", "a"])
huruf.discard("x")
diambil = huruf.pop()

print("Set akhir:", huruf)
print("Item pop:", diambil)
```

## 8.6. Operasi Himpunan (Union, Intersection, Difference)
Kekuatan utama set ada di operasi himpunan. `union` menggabungkan semua nilai unik, `intersection` mengambil nilai yang sama, dan `difference` mengambil nilai yang hanya ada di salah satu set. Operasi ini membantu membandingkan dua kelompok data dengan cepat. Sangat berguna untuk analisis data dasar.


```python
# Operasi himpunan
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference a-b:", a - b)
print("Symmetric difference:", a ^ b)
```

## 8.7. Latihan

- Buat dua set hobi dan tampilkan hobi yang sama.

## 8.8. Ringkasan
Set adalah struktur data unik untuk mengelola nilai tanpa duplikasi. Set bersifat mutable, tidak indexable, dan sangat kuat untuk operasi perbandingan kelompok data.
