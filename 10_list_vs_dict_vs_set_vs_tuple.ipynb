## 10.2 Ringkasan Konsep
- `list`: urutan data, bisa duplikat, bisa diubah (`mutable`), akses via index.
- `dict`: pasangan `key: value`, key unik, bisa diubah, akses via key.
- `set`: kumpulan unik tanpa urutan index, bisa diubah, cocok untuk operasi himpunan.
- `tuple`: urutan data, bisa duplikat, tidak bisa diubah (`immutable`).


## 10.3 Tabel Perbandingan Cepat
| Struktur | Berurutan | Bisa Duplikat | Bisa Diubah | Cara Akses Utama |
|---|---|---|---|---|
| list | Ya | Ya | Ya | index |
| dict | Ya (Python modern mempertahankan urutan input) | Key: tidak | Ya | key |
| set | Tidak menjamin urutan tampilan | Tidak | Ya | iterasi / membership |
| tuple | Ya | Ya | Tidak | index |


```python
# Contoh bentuk data
list_data = [10, 20, 20, 30]
dict_data = {"nama": "Rani", "umur": 21}
set_data = {"apel", "jeruk", "apel"}
tuple_data = ("A", "B", "B")

print('list  :', list_data)
print('dict  :', dict_data)
print('set   :', set_data)
print('tuple :', tuple_data)
```

## 10.4 Kapan Pakai yang Mana?
- Pakai `list` saat data perlu urutan dan sering berubah.
- Pakai `dict` saat data punya label/atribut (misal biodata, konfigurasi).
- Pakai `set` saat fokus ke data unik dan cek keanggotaan cepat.
- Pakai `tuple` saat data seharusnya tetap (misal koordinat, konstanta).


```python
# Studi kasus singkat
nama_peserta = ["Ana", "Budi", "Ana", "Caca"]
unik_peserta = set(nama_peserta)
profil = {"nama": "Ana", "nilai": 88}
koordinat = (-6.2, 106.8)

print('Jumlah peserta unik:', len(unik_peserta))
print('Nilai Ana:', profil['nilai'])
print('Koordinat:', koordinat)
```

## 10.5 Kesalahan Umum Pemula
- Memakai `list` untuk data berlabel, padahal lebih jelas pakai `dict`.
- Menyimpan data yang seharusnya unik dalam `list`, padahal lebih tepat `set`.
- Mengubah data tetap dengan `list`, padahal semestinya `tuple` supaya aman.
