## Tujuan Belajar

## 1) Dasar Tuple


**Penjelasan Code 1:** Sel ini membuat tuple dan menampilkan tipe datanya.


```python
# Menampilkan tipe data dari beberapa nilai.

koordinat = (10, 20)
print(koordinat)
print(type(koordinat))
```

## 2) Tuple Immutable


**Penjelasan Code 2:** Sel ini menegaskan tuple immutable (tidak bisa diubah setelah dibuat).


```python
# koordinat[0] = 99  # Error, tuple tidak bisa diubah
print("Tuple tidak bisa diubah setelah dibuat.")
```

## 3) Unpacking Tuple


**Penjelasan Code 3:** Sel ini melakukan unpacking tuple ke beberapa variabel.


```python
# Sel ini melakukan unpacking tuple ke beberapa variabel.

x, y = koordinat
print("x:", x)
print("y:", y)
```

## 4) Sort Dictionary Berdasarkan Value


**Penjelasan Code 4:** Sel ini mengurutkan data dictionary berdasarkan value dengan bantuan tuple.


```python
# Perulangan for menelusuri item satu per satu.

skor = {"Ani": 88, "Budi": 95, "Cici": 90}

pasangan = []
for nama, nilai in skor.items():
    pasangan.append((nilai, nama))

pasangan.sort(reverse=True)
print(pasangan)
```

## 5) List Comprehension Singkat


**Penjelasan Code 5:** Sel ini mengenalkan list comprehension untuk menulis loop sederhana lebih ringkas.


```python
# Perulangan for menelusuri item satu per satu.

angka = [1, 2, 3, 4, 5]
kuadrat = [n*n for n in angka]
print(kuadrat)
```

## Latihan
- Simpan data `nama, usia` ke dalam tuple.
- Dari dictionary produk-harga, tampilkan 3 harga tertinggi.


## Ringkasan
Tuple cocok untuk data yang tidak perlu diubah, dan sering dipakai saat bekerja dengan dictionary.
