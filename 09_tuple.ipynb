## 9.2. Dasar Struktur: Cara Mendefinisikan Tuple
Tuple mirip list, tapi datanya tidak bisa diubah setelah dibuat. Kamu bisa membuat tuple dengan tanda kurung `()` atau fungsi `tuple()`. Satu item tuple perlu koma, misalnya `(10,)`, supaya Python tahu itu tuple, bukan angka biasa. Tuple cocok untuk data tetap seperti koordinat atau pasangan nilai.


```python
# Cara mendefinisikan tuple
t1 = (10, 20, 30)
t2 = tuple(["a", "b", "c"])
t3 = (99,)  # tuple satu elemen

print(t1)
print(t2)
print(t3, type(t3))
```

## 9.3. Sifat Tuple (`immutable`, `indexable`, `duplicable`)
Tuple bersifat `immutable`, jadi elemen tidak bisa diubah langsung. Tuple tetap `indexable`, jadi bisa diakses dengan index positif/negatif. Tuple juga `duplicable`, sehingga bisa diulang dengan operator `*`. Karena immutable, tuple relatif aman untuk data yang tidak boleh berubah.


```python
# Sifat tuple
koordinat = (10, 20)
print("Indexable:", koordinat[0], koordinat[-1])
print("Duplicable:", koordinat * 2)

# koordinat[0] = 99  # error: tuple immutable
```

## 9.4. Cara Akses Tuple (Index dan Slicing)
Akses tuple mirip list dan string: bisa satu elemen, range, step, atau dibalik. Karena immutable, slicing tuple akan menghasilkan tuple baru, bukan mengubah tuple lama. Ini penting saat kamu ingin ambil sebagian data tanpa merusak data asli. Gunakan pola akses ini untuk membaca data secara aman.


```python
# Variasi akses tuple
data = (0, 1, 2, 3, 4, 5)
print(data[2])
print(data[-1])
print(data[1:4])
print(data[:3])
print(data[3:])
print(data[::2])
print(data[::-1])
```

## 9.5. Fungsi/Method Penting pada Tuple
Tuple punya method bawaan yang sederhana: `count()` dan `index()`. Untuk proses lain, biasanya kita pakai fungsi umum Python seperti `len()`, `min()`, `max()`, dan `sum()` (kalau isi angka). Jadi walau method tuple sedikit, tuple tetap sangat berguna. Kuncinya ada pada sifat immutable dan kemudahan aksesnya.


```python
# Method/fungsi penting untuk tuple
nilai = (80, 90, 75, 90)
print(nilai.count(90))
print(nilai.index(75))
print(len(nilai))
print(min(nilai), max(nilai), sum(nilai))
```

## 9.6. Operasi Tuple yang Umum
Operasi tuple mirip sequence lain: gabung dengan `+`, ulang dengan `*`, cek keberadaan item dengan `in`, dan unpacking ke beberapa variabel. Unpacking sering dipakai agar kode lebih ringkas dan mudah dibaca. Ini salah satu pola tuple yang paling berguna di kehidupan nyata. Biasakan praktikkan di data sederhana dulu.


```python
# Operasi tuple umum + unpacking
a = (1, 2)
b = (3, 4)
print(a + b)
print(a * 2)
print(2 in a)

x, y = a
print(x, y)
```
