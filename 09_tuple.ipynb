## 1) Dasar Tuple


Di sel ini kita berkenalan dengan tuple, yaitu struktur mirip list tetapi tidak bisa diubah. Tuple cocok untuk data yang sifatnya tetap, seperti koordinat atau pasangan nilai konstan. Mengetahui perbedaan tuple dan list akan membantu kamu memilih struktur data yang tepat. Ini keputusan kecil yang berdampak pada desain program.


```python
# Menampilkan tipe data dari beberapa nilai.

koordinat = (10, 20)
print(koordinat)
print(type(koordinat))
```

## 2) Tuple Immutable


Bagian ini menegaskan sifat immutable dari tuple. Saat kamu mencoba mengubah elemennya, Python akan menolak. Justru di situlah manfaat tuple: data lebih aman dari perubahan tidak sengaja. Jadi pakailah tuple ketika kamu ingin menjaga data tetap konsisten.


```python
# koordinat[0] = 99  # Error, tuple tidak bisa diubah
print("Tuple tidak bisa diubah setelah dibuat.")
```

## 3) Unpacking Tuple


Sel ini memperlihatkan teknik unpacking, yaitu membongkar isi tuple langsung ke beberapa variabel. Cara ini membuat kode lebih ringkas dan enak dibaca. Kamu akan sering melihat pola ini saat fungsi mengembalikan lebih dari satu nilai. Setelah terbiasa, unpacking terasa sangat natural.


```python
# Sel ini melakukan unpacking tuple ke beberapa variabel.

x, y = koordinat
print("x:", x)
print("y:", y)
```

## 4) Sort Dictionary Berdasarkan Value


Di sini kita kombinasikan dictionary dan tuple untuk proses pengurutan berdasarkan nilai. Kita ubah pasangan `(nama, skor)` menjadi `(skor, nama)` agar sorting mengikuti skor terlebih dahulu. Teknik ini sering dipakai saat membuat ranking. Contoh ini menunjukkan bagaimana struktur data bisa saling melengkapi.


```python
# Perulangan for menelusuri item satu per satu.

skor = {"Ani": 88, "Budi": 95, "Cici": 90}

pasangan = []
for nama, nilai in skor.items():
    pasangan.append((nilai, nama))

pasangan.sort(reverse=True)
print(pasangan)
```

## 5) List Comprehension Singkat


Bagian terakhir mengenalkan list comprehension sebagai cara ringkas menulis loop sederhana. Walau singkat, idenya tetap sama: memproses tiap item dan membentuk list baru. Untuk pemula, fokus dulu pada keterbacaan sebelum mengejar gaya paling pendek. Kalau logikanya sudah paham, comprehension akan jadi alat yang sangat powerful.


```python
# Perulangan for menelusuri item satu per satu.

angka = [1, 2, 3, 4, 5]
kuadrat = [n*n for n in angka]
print(kuadrat)
```
