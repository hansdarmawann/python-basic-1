# Bab 8: List


## Tujuan Belajar
- Membuat dan memodifikasi list.
- Menelusuri list dengan loop.
- Menggunakan method list (`append`, `sort`, `pop`).
- Mengubah string menjadi list dan sebaliknya.


## 1) Dasar List


```python
buah = ["apel", "jeruk", "mangga"]
print(buah)
print("Item pertama:", buah[0])
```

## 2) List Bersifat Mutable


```python
buah = ["apel", "jeruk", "mangga"]
buah[1] = "pisang"
print(buah)
```

## 3) Method Penting


```python
angka = [5, 2, 9]
angka.append(7)
angka.sort()
print(angka)
angka.pop()
print(angka)
```

## 4) Looping List


```python
nilai = [80, 75, 90, 60]
for n in nilai:
    if n >= 75:
        print(n, "Lulus")
    else:
        print(n, "Belum lulus")
```

## 5) String <-> List


```python
kalimat = "saya belajar python"
kata = kalimat.split()
print(kata)
print("-".join(kata))
```

## Latihan
- Dari list angka, cari nilai minimum tanpa `min()`.
- Buat list belanja, lalu hapus item yang sudah dibeli.


## Ringkasan
List cocok untuk menyimpan kumpulan data berurutan yang bisa diubah.
