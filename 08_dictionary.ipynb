## Tujuan Belajar
- Memahami pasangan `key: value`.
- Menambah, mengubah, dan membaca data dictionary.
- Menghitung frekuensi data sederhana.
- Iterasi dictionary dengan `items()`.


## 1) Dasar Dictionary


**Penjelasan Code 1:** Sel ini membuat dictionary sederhana dengan pasangan `key: value`.


```python
# Sel ini membuat dictionary sederhana dengan pasangan `key: value`.

mahasiswa = {
    "nama": "Dina",
    "umur": 21,
    "jurusan": "Informatika"
}

print(mahasiswa)
print(mahasiswa["nama"])
```

## 2) Update dan Tambah Data


**Penjelasan Code 2:** Sel ini mengubah nilai lama dan menambah key baru dalam dictionary.


```python
# Sel ini mengubah nilai lama dan menambah key baru dalam dictionary.

mahasiswa["umur"] = 22
mahasiswa["kota"] = "Bandung"
print(mahasiswa)
```

## 3) Frekuensi Kata


**Penjelasan Code 3:** Sel ini menghitung frekuensi kata menggunakan pola `dict.get()`.


```python
# Perulangan for menelusuri item satu per satu.

teks = "python data python coding data python"
frekuensi = {}

for kata in teks.split():
    frekuensi[kata] = frekuensi.get(kata, 0) + 1

print(frekuensi)
```

## 4) Looping Dictionary


**Penjelasan Code 4:** Sel ini menelusuri dictionary dengan `items()` untuk mengambil key dan value.


```python
# Perulangan for menelusuri item satu per satu.

for k, v in frekuensi.items():
    print(k, "->", v)
```

## 5) Cek Ketersediaan Key


**Penjelasan Code 5:** Sel ini mengecek apakah suatu key ada sebelum mengakses nilainya.


```python
# Sel ini mengecek apakah suatu key ada sebelum mengakses nilainya.

if "python" in frekuensi:
    print("Jumlah kata 'python':", frekuensi["python"])
```

## Latihan
- Buat dictionary biodata sederhana kamu.
- Hitung frekuensi huruf dari sebuah kata (contoh: `banana`).


## Ringkasan
Dictionary sangat efektif untuk data yang berbasis pasangan kunci dan nilai.
