## 7.2. Dasar Struktur: Cara Mendefinisikan Dictionary
Dictionary menyimpan data dalam pasangan `key: value`. Kita bisa menulis langsung dengan kurung kurawal atau pakai fungsi `dict()`. Dictionary cocok saat data punya label jelas, misalnya `nama`, `umur`, atau `kota`. Struktur ini sangat sering dipakai di aplikasi dan API.


```python
# Cara mendefinisikan dictionary
d1 = {"nama": "Dina", "umur": 21}
d2 = dict(kota="Bandung", aktif=True)
d3 = dict([("jurusan", "Informatika"), ("semester", 4)])

print(d1)
print(d2)
print(d3)
```

## 7.3. Sifat Dictionary (`mutable`, key unik, akses by key)
Dictionary bersifat `mutable`, jadi data bisa diubah/diperbarui. Dictionary **bukan indexable secara urutan angka** seperti list; aksesnya memakai `key`. Key harus unik, jadi jika key sama ditulis lagi, nilai lama akan tertimpa. Value boleh duplikat, itu tidak masalah.


```python
# Sifat dictionary
profil = {"nama": "Rina", "umur": 20, "umur": 21}  # key 'umur' tertimpa
profil["kota"] = "Jakarta"                              # mutable

print(profil)
print("Akses by key:", profil["nama"])
```

## 7.4. Cara Akses dan Iterasi Dictionary
Akses dictionary bisa pakai `[]` atau `.get()`. Bedanya, `.get()` lebih aman karena tidak error jika key tidak ada. Untuk menelusuri isi dictionary, pakai `keys()`, `values()`, atau `items()`. Pola ini penting untuk olah data berbasis pasangan key-value.


```python
# Akses dan iterasi dictionary
mahasiswa = {"nama": "Budi", "umur": 22, "kota": "Solo"}

print(mahasiswa["nama"])
print(mahasiswa.get("ipk"))

for k, v in mahasiswa.items():
    print(k, "->", v)
```

## 7.5. Method Dictionary Penting + Penjelasan
Method penting dictionary yang perlu kamu kuasai:
- `get(key, default)` ambil nilai key dengan aman.
- `update(dict_baru)` memperbarui banyak data sekaligus.
- `pop(key)` menghapus key tertentu.
- `keys()`, `values()`, `items()` untuk iterasi isi dictionary.
- `clear()` mengosongkan dictionary.


```python
# Method dictionary penting
data = {"apel": 3, "jeruk": 5}
print(data.get("apel", 0))

data.update({"mangga": 2, "apel": 4})
hapus = data.pop("jeruk")

print("Data:", data)
print("Yang dihapus:", hapus)
print("Keys:", list(data.keys()))
print("Values:", list(data.values()))
print("Items:", list(data.items()))
```

## 7.6. Operasi Dictionary yang Umum
Operasi umum dictionary meliputi cek keberadaan key dengan `in`, menghitung jumlah pasangan dengan `len()`, dan menghapus data dengan `del`. Untuk kasus hitung frekuensi, pola `dict.get(..., 0) + 1` sangat populer. Ini pattern penting yang akan sering muncul di soal interview dasar. Latih sampai terasa otomatis.


```python
# Operasi dictionary umum
teks = "python data python"
freq = {}
for kata in teks.split():
    freq[kata] = freq.get(kata, 0) + 1

print(freq)
print("python" in freq)
print(len(freq))
```
