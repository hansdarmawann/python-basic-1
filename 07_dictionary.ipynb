## Tujuan Belajar
- Memahami pasangan `key: value`.
- Menambah, mengubah, dan membaca data dictionary.

## 1) Dasar Dictionary


Sel ini memperkenalkan dictionary sebagai pasangan `key: value`. Berbeda dari list yang pakai indeks angka, dictionary diakses lewat kunci yang bermakna. Karena itu, struktur ini cocok untuk data seperti profil, konfigurasi, atau mapping informasi. Coba lihat bagaimana `nama` dipanggil langsung dari key.


```python
# Sel ini membuat dictionary sederhana dengan pasangan `key: value`.

mahasiswa = {
    "nama": "Dina",
    "umur": 21,
    "jurusan": "Informatika"
}

print(mahasiswa)
print(mahasiswa["nama"])
```

## 2) Update dan Tambah Data


Di bagian ini kita update data lama dan menambah data baru. Salah satu kelebihan dictionary adalah proses ini bisa dilakukan dengan sangat langsung. Pola ini sering dipakai untuk memperbarui data pengguna atau status proses. Praktis dan mudah dibaca saat kode makin panjang.


```python
# Sel ini mengubah nilai lama dan menambah key baru dalam dictionary.

mahasiswa["umur"] = 22
mahasiswa["kota"] = "Bandung"
print(mahasiswa)
```

## 3) Frekuensi Kata


Sel ini mengajarkan pola hitung frekuensi yang sangat populer. Kita loop setiap kata lalu menambah hitungan dengan bantuan `get()`. Teknik ini berguna untuk analisis teks, statistik sederhana, sampai pra-pemrosesan data. Ini salah satu pattern Python yang wajib kamu kuasai.


```python
# Perulangan for menelusuri item satu per satu.

teks = "python data python coding data python"
frekuensi = {}

for kata in teks.split():
    frekuensi[kata] = frekuensi.get(kata, 0) + 1

print(frekuensi)
```

## 4) Looping Dictionary


Di sini kita menelusuri isi dictionary dengan `items()`. Dengan cara ini, key dan value bisa diakses bersamaan dalam satu loop. Hasilnya lebih rapi dibanding mengambil key lalu mencari value satu per satu. Ini pendekatan standar saat kamu ingin menampilkan seluruh isi dictionary.


```python
# Perulangan for menelusuri item satu per satu.

for k, v in frekuensi.items():
    print(k, "->", v)
```

## 5) Cek Ketersediaan Key


Bagian ini menunjukkan cara aman mengakses key dengan pengecekan terlebih dulu. Kenapa penting? Karena akses key yang tidak ada bisa memicu error. Dengan `if key in dict`, program jadi lebih robust. Ini kebiasaan kecil yang menghindari banyak masalah di kemudian hari.


```python
# Sel ini mengecek apakah suatu key ada sebelum mengakses nilainya.

if "python" in frekuensi:
    print("Jumlah kata 'python':", frekuensi["python"])
```

Dictionary sangat efektif untuk data yang berbasis pasangan kunci dan nilai.
