## Tujuan Belajar
- Membuat dan memodifikasi list.
- Menelusuri list dengan loop.
- Menggunakan method list (`append`, `sort`, `pop`).
- Mengubah string menjadi list dan sebaliknya.


## 1) Dasar List


**Penjelasan Code 1:** Sel ini membuat list dan mengakses elemen berdasarkan indeks.


```python
# Sel ini membuat list dan mengakses elemen berdasarkan indeks.

buah = ["apel", "jeruk", "mangga"]
print(buah)
print("Item pertama:", buah[0])
```

## 2) List Bersifat Mutable


**Penjelasan Code 2:** Sel ini menunjukkan list bersifat mutable sehingga elemennya bisa diubah.


```python
# Sel ini menunjukkan list bersifat mutable sehingga elemennya bisa diubah.

buah = ["apel", "jeruk", "mangga"]
buah[1] = "pisang"
print(buah)
```

## 3) Method Penting


**Penjelasan Code 3:** Sel ini mencontohkan method list: tambah data, urutkan, lalu hapus item terakhir.


```python
# Sel ini mencontohkan method list: tambah data, urutkan, lalu hapus item terakhir.

angka = [5, 2, 9]
angka.append(7)
angka.sort()
print(angka)
angka.pop()
print(angka)
```

## 4) Looping List


**Penjelasan Code 4:** Sel ini menelusuri list nilai lalu memberi label lulus/belum lulus.


```python
# Perulangan for menelusuri item satu per satu.

nilai = [80, 75, 90, 60]
for n in nilai:
    if n >= 75:
        print(n, "Lulus")
    else:
        print(n, "Belum lulus")
```

## 5) String <-> List


**Penjelasan Code 5:** Sel ini mengubah string ke list dengan `split()` lalu gabung lagi dengan `join()`.


```python
# Mengubah format data teks dan list.

kalimat = "saya belajar python"
kata = kalimat.split()
print(kata)
print("-".join(kata))
```

## Latihan
