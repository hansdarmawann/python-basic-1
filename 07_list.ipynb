## 1) Dasar List


Di sel ini kita mulai dari list sebagai wadah data berurutan. Kamu bisa menyimpan banyak nilai dalam satu variabel dan mengaksesnya lewat indeks. Konsep ini akan sering muncul saat mengolah kumpulan data. Coba ubah isinya sendiri untuk melihat cara kerjanya.


```python
# Sel ini membuat list dan mengakses elemen berdasarkan indeks.

buah = ["apel", "jeruk", "mangga"]
print(buah)
print("Item pertama:", buah[0])
```

## 2) List Bersifat Mutable


Bagian ini menegaskan bahwa list bersifat mutable, artinya elemennya bisa diubah. Ini berbeda dengan string yang tidak bisa diubah langsung. Karena fleksibel, list cocok untuk data yang dinamis. Contoh ini sederhana tapi sangat mewakili penggunaan nyata.


```python
# Sel ini menunjukkan list bersifat mutable sehingga elemennya bisa diubah.

buah = ["apel", "jeruk", "mangga"]
buah[1] = "pisang"
print(buah)
```

## 3) Method Penting


Sel ini memperkenalkan method list yang paling sering dipakai. `append()` menambah item, `sort()` mengurutkan, dan `pop()` menghapus item terakhir. Dengan tiga method ini saja, kamu sudah bisa melakukan banyak operasi data dasar. Biasakan baca hasil setiap langkah supaya alurnya jelas.


```python
# Sel ini mencontohkan method list: tambah data, urutkan, lalu hapus item terakhir.

angka = [5, 2, 9]
angka.append(7)
angka.sort()
print(angka)
angka.pop()
print(angka)
```

## 4) Looping List


Di sini kita gabungkan list dengan percabangan dalam loop. Setiap nilai dicek lalu diberi label lulus atau belum lulus. Pola ini umum pada validasi dan klasifikasi sederhana. Dari contoh ini, kamu bisa lanjut ke kasus seperti filtering data berdasarkan kriteria.


```python
# Perulangan for menelusuri item satu per satu.

nilai = [80, 75, 90, 60]
for n in nilai:
    if n >= 75:
        print(n, "Lulus")
    else:
        print(n, "Belum lulus")
```

## 5) String <-> List


Bagian akhir menunjukkan hubungan string dan list. `split()` memecah kalimat menjadi list kata, sedangkan `join()` menggabungkannya kembali. Ini teknik yang sangat sering dipakai saat olah data teks. Kalau kamu nyaman dengan ini, pekerjaan parsing akan jadi jauh lebih mudah.


```python
# Mengubah format data teks dan list.

kalimat = "saya belajar python"
kata = kalimat.split()
print(kata)
print("-".join(kata))
```
