## 1) Membuka dan Membaca File


Sel ini menunjukkan cara membaca seluruh isi file sekaligus menggunakan `.read()`. Metode ini cocok untuk file kecil sampai menengah yang isinya ingin langsung diproses penuh. Kita juga pakai `with open(...)` agar file tertutup otomatis setelah selesai. Kebiasaan ini penting supaya pengelolaan resource tetap aman.


```python
# Pastikan file contoh.txt ada di folder yang sama
with open("contoh.txt", "r", encoding="utf-8") as f:
    isi = f.read()

print(isi)
```

## 2) Membaca per Baris


Di sini kita baca file baris demi baris menggunakan loop. Pendekatan ini lebih hemat memori, terutama jika ukuran file besar. Setiap baris diproses satu per satu, jadi cocok untuk analisis log atau data teks panjang. `strip()` dipakai agar hasil tampil lebih bersih.


```python
# Perulangan for menelusuri item satu per satu.

with open("contoh.txt", "r", encoding="utf-8") as f:
    for baris in f:
        print(baris.strip())
```

## 3) Mencari Kata di File


Bagian ini mencontohkan pencarian kata kunci di dalam file. Kita ubah baris menjadi huruf kecil dulu supaya pencarian tidak sensitif huruf besar-kecil. Pola ini sering dipakai saat filtering data teks. Dengan teknik sederhana ini, kamu sudah bisa membuat mini-search di file.


```python
# Perulangan for menelusuri item satu per satu.

kata_kunci = "python"
with open("contoh.txt", "r", encoding="utf-8") as f:
    for baris in f:
        if kata_kunci in baris.lower():
            print("Ditemukan:", baris.strip())
```

## 4) Menulis File


Sel ini memperlihatkan cara menulis file baru dengan mode `w`. Kalau file sudah ada, isinya akan ditimpa, jadi kamu perlu sadar efeknya. Kita menulis teks per baris agar hasilnya rapi dan mudah dibaca. Ini dasar penting untuk menyimpan output program secara permanen.


```python
# Membuka file mode tulis dan menimpa isi lama jika ada.

with open("hasil.txt", "w", encoding="utf-8") as f:
    f.write("Ini file baru.
")
    f.write("Belajar file di Python itu penting.
")

print("File hasil.txt berhasil dibuat.")
```

## 5) try-except saat Buka File


Terakhir, kita tangani error saat file tidak ditemukan. Daripada program berhenti dengan pesan teknis yang membingungkan, kita tampilkan pesan yang lebih manusiawi. `try-except` sangat membantu ketika program berinteraksi dengan file dari pengguna. Ini langkah kecil yang membuat aplikasi terasa jauh lebih profesional.


```python
# Menangani error agar program tidak langsung berhenti.

try:
    with open("tidak_ada.txt", "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print("File tidak ditemukan. Periksa nama file.")
```
