## Tujuan Belajar
- Membaca file teks.
- Menelusuri isi file baris per baris.
- Menulis file baru.
- Menangani error saat file tidak ditemukan.


## 1) Membuka dan Membaca File


**Penjelasan Code 1:** Sel ini membaca seluruh isi file sekaligus dengan `.read()`.


```python
# Pastikan file contoh.txt ada di folder yang sama
with open("contoh.txt", "r", encoding="utf-8") as f:
    isi = f.read()

print(isi)
```

## 2) Membaca per Baris


**Penjelasan Code 2:** Sel ini membaca file baris per baris agar lebih hemat memori untuk file besar.


```python
# Perulangan for menelusuri item satu per satu.

with open("contoh.txt", "r", encoding="utf-8") as f:
    for baris in f:
        print(baris.strip())
```

## 3) Mencari Kata di File


**Penjelasan Code 3:** Sel ini mencari kata tertentu dari setiap baris file.


```python
# Perulangan for menelusuri item satu per satu.

kata_kunci = "python"
with open("contoh.txt", "r", encoding="utf-8") as f:
    for baris in f:
        if kata_kunci in baris.lower():
            print("Ditemukan:", baris.strip())
```

## 4) Menulis File


**Penjelasan Code 4:** Sel ini membuat file baru dan menulis beberapa baris teks.


```python
# Membuka file mode tulis dan menimpa isi lama jika ada.

with open("hasil.txt", "w", encoding="utf-8") as f:
    f.write("Ini file baru.
")
    f.write("Belajar file di Python itu penting.
")

print("File hasil.txt berhasil dibuat.")
```

## 5) try-except saat Buka File


**Penjelasan Code 5:** Sel ini menangani kondisi file tidak ditemukan dengan `try-except`.


```python
# Menangani error agar program tidak langsung berhenti.

try:
    with open("tidak_ada.txt", "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print("File tidak ditemukan. Periksa nama file.")
```

## Latihan
- Hitung jumlah baris dalam sebuah file.
- Simpan semua baris yang mengandung kata tertentu ke file baru.
