## Tujuan Belajar
- Memahami `while` dan `for`.
- Menggunakan `break` dan `continue`.
- Mengenal pola umum perulangan (hitung, jumlah, max/min).


## 1) while


**Penjelasan Code 1:** Sel ini memakai `while` untuk perulangan dengan kondisi berhenti.


```python
# Perulangan while berjalan selama kondisi bernilai True.

angka = 1
while angka <= 5:
    print("Angka:", angka)
    angka += 1
```

## 2) for


**Penjelasan Code 2:** Sel ini memakai `for` untuk menelusuri setiap item dalam list.


```python
# Perulangan for menelusuri item satu per satu.

for nama in ["Ani", "Budi", "Cici"]:
    print("Halo", nama)
```

## 3) break dan continue


**Penjelasan Code 3:** Sel ini menunjukkan perbedaan `break` (hentikan loop) dan `continue` (lewati iterasi ini).


```python
# Perulangan for menelusuri item satu per satu.

for i in range(1, 8):
    if i == 6:
        break
    if i % 2 == 0:
        continue
    print(i)
```

## 4) Pola Menjumlahkan


**Penjelasan Code 4:** Sel ini menunjukkan pola akumulasi/jumlah total di dalam loop.


```python
# Perulangan for menelusuri item satu per satu.

total = 0
for n in [10, 20, 30]:
    total += n
print("Total:", total)
```

## 5) Pola Cari Nilai Maksimum


**Penjelasan Code 5:** Sel ini menunjukkan pola mencari nilai maksimum secara manual.


```python
# Perulangan for menelusuri item satu per satu.

data = [8, 3, 15, 6, 20, 1]
terbesar = data[0]
for n in data:
    if n > terbesar:
        terbesar = n
print("Nilai terbesar:", terbesar)
```
