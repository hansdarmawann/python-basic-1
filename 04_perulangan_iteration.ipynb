## 4.1. Tujuan Belajar

- Menggunakan `break` dan `continue`.

## 4.2. while


Sel ini menunjukkan `while`, yaitu loop yang berjalan selama kondisi masih benar. Model ini cocok saat kita belum tahu pasti jumlah pengulangan dari awal. Yang penting, jangan lupa membuat kondisi berhenti agar tidak jadi loop tak berujung. Di contoh ini, variabel dinaikkan satu per satu sampai batas tertentu.


```python
# Perulangan while berjalan selama kondisi bernilai True.

angka = 1
while angka <= 5:
    print("Angka:", angka)
    angka += 1
```

## 4.3. for


Di bagian ini kita pakai `for` untuk menelusuri item dalam list. Pendekatan `for` biasanya lebih ringkas dan aman saat jumlah data sudah jelas. Setiap putaran, variabel `nama` akan berisi satu item dari list. Ini cara yang sangat umum untuk memproses kumpulan data.


```python
# Perulangan for menelusuri item satu per satu.

for nama in ["Ani", "Budi", "Cici"]:
    print("Halo", nama)
```

## 4.4. break dan continue


Sel ini memperlihatkan dua kontrol penting dalam loop: `break` dan `continue`. `break` langsung menghentikan loop, sedangkan `continue` melewati sisa proses di iterasi saat ini. Dua alat ini berguna banget ketika kita ingin menyaring kondisi tertentu di tengah perulangan. Coba eksperimen dengan mengubah syaratnya supaya kamu benar-benar paham bedanya.


```python
# Perulangan for menelusuri item satu per satu.

for i in range(1, 8):
    if i == 6:
        break
    if i % 2 == 0:
        continue
    print(i)
```

## 4.5. Pola Menjumlahkan


Di sini kita belajar pola akumulasi, yaitu menjumlahkan nilai sedikit demi sedikit dalam loop. Biasanya kita mulai dari `total = 0`, lalu tambahkan setiap data yang lewat. Pola ini sering dipakai untuk menghitung total belanja, total nilai, dan banyak kasus lain. Konsepnya sederhana, tapi sangat fundamental.


```python
# Perulangan for menelusuri item satu per satu.

total = 0
for n in [10, 20, 30]:
    total += n
print("Total:", total)
```

## 4.6. Pola Cari Nilai Maksimum


Sel terakhir mengenalkan pola mencari nilai maksimum secara manual. Kita mulai dari asumsi nilai pertama sebagai yang terbesar, lalu bandingkan dengan elemen berikutnya. Jika ketemu nilai lebih besar, variabel maksimum diperbarui. Teknik ini membantu kamu memahami logika kerja fungsi `max()` dari dalam.


```python
# Perulangan for menelusuri item satu per satu.

data = [8, 3, 15, 6, 20, 1]
terbesar = data[0]
for n in data:
    if n > terbesar:
        terbesar = n
print("Nilai terbesar:", terbesar)
```
