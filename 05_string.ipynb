## 1) Sifat Dasar String
String adalah data teks, dan di Python string diperlakukan sebagai urutan karakter. Artinya kita bisa ambil panjangnya, mengakses per karakter dengan indeks, dan melakukan slicing. Hal penting yang harus diingat: string itu `immutable`, jadi karakter di posisi tertentu tidak bisa diubah langsung. Kalau ingin mengubah isi, kita membuat string baru.


Di kode ini kita lihat contoh panjang string, akses karakter, dan indeks negatif. Indeks negatif sangat berguna saat kita ingin ambil karakter dari belakang tanpa menghitung panjang dulu. Konsep kecil ini nanti sering muncul saat parsing data teks. Jalankan perlahan sambil ubah isi teksnya supaya terasa bedanya.


```python
# String adalah urutan karakter (sequence)
teks = "Belajar Python"
print(teks)
print("Panjang:", len(teks))
print("Karakter pertama:", teks[0])
print("Karakter terakhir:", teks[-1])
```

## 2) Operasi String yang Umum
String mendukung beberapa operasi penting yang sering dipakai. Kita bisa menggabungkan string dengan `+`, mengulang string dengan `*`, mengecek keberadaan substring dengan `in`, serta membandingkan dua string. Selain itu, slicing membantu mengambil bagian teks tertentu dengan cepat. Ini semua adalah toolkit dasar untuk olah teks.


Perhatikan contoh berikut karena ini kombinasi operasi yang paling sering dipakai pemula sampai intermediate. Dari sisi praktik, operasi-operasi ini sering dipakai di validasi input, formatting output, dan ekstraksi data. Kalau ingin cepat mahir string, fokus kuasai bagian ini dulu. Sisanya tinggal pembiasaan.


```python
# Operasi string: gabung, ulang, cek isi, slicing, dan perbandingan
nama_depan = "Budi"
nama_belakang = "Santoso"
nama_lengkap = nama_depan + " " + nama_belakang

print("Gabung:", nama_lengkap)
print("Ulang:", "ha" * 3)
print("Cek substring 'Python' ada?:", "Python" in "Belajar Python Dasar")
print("Slicing [0:7]:", "Belajar Python"[0:7])
print("Perbandingan 'apel' < 'jeruk':", "apel" < "jeruk")
```

## 3) String Itu Immutable
Banyak pemula mencoba mengganti satu karakter string langsung lewat indeks dan kena error. Itu normal, karena sifat string memang immutable. Solusinya bukan memaksa ubah karakter lama, tapi membentuk string baru dari bagian yang dibutuhkan. Pola ini aman dan jadi standar penulisan Python.


Contoh ini sengaja sederhana supaya idenya jelas: karakter pertama kita ganti lewat teknik gabung. Nanti pola yang sama bisa dipakai untuk kasus lebih kompleks seperti normalisasi teks atau pembersihan data user. Intinya, berpikirnya bukan “edit string lama”, tapi “bangun string baru”.


```python
# String immutable: buat string baru saat ingin "mengubah"
nama = "andi"
# nama[0] = "A"  # Ini akan error
nama_baru = "A" + nama[1:]
print(nama_baru)
```

## 4) Method String yang Paling Sering Dipakai
Python punya banyak method string, tapi tidak perlu hafal semua di awal. Fokus dulu ke method yang paling sering muncul: `strip()`, `lower()`, `upper()`, `replace()`, `split()`, `join()`, `startswith()`, dan `endswith()`. Dengan set kecil ini saja, kamu sudah bisa menyelesaikan banyak tugas olah teks sehari-hari. Nanti kalau butuh lebih, baru eksplor method tambahan.


Di kode ini saya rangkum method-method populer dalam satu tempat supaya mudah jadi referensi cepat. Coba ubah isi kalimat, pemisah, atau kata yang diganti agar kamu paham perilaku tiap method. Latihan kecil seperti ini jauh lebih efektif daripada menghafal teori. Yang penting kamu ngerti kapan method tertentu dipakai.


```python
# Method string yang umum dipakai
kalimat = "  saya suka Python  "

print("strip():", kalimat.strip())
print("lower():", kalimat.lower())
print("upper():", kalimat.upper())
print("replace():", kalimat.replace("Python", "data"))
print("split():", kalimat.strip().split())
print("join():", "-".join(["python", "basic", "id"]))
print("startswith('  sa'):", kalimat.startswith("  sa"))
print("endswith('  '):", kalimat.endswith("  "))
```

## 5) f-string untuk Format Output
Setelah mengolah string, kita biasanya perlu menampilkan hasil ke pengguna. f-string adalah cara modern yang rapi dan mudah dibaca untuk menyisipkan variabel ke dalam teks. Dibanding gabung string manual, f-string mengurangi potensi error dan lebih nyaman dirawat. Praktikkan ini sejak awal supaya gaya coding kamu konsisten.


Contoh berikut menunjukkan format teks yang natural dan mudah dipahami orang awam. Teknik ini sangat berguna untuk membuat pesan aplikasi, laporan sederhana, dan output hasil analisis. Kamu juga bisa menambahkan format angka di f-string saat nanti belajar topik lanjutan. Untuk sekarang, pahami dulu alur dasarnya.


```python
# f-string untuk menyisipkan variabel ke dalam teks
nama = "Rani"
umur = 22
print(f"Nama saya {nama}, umur saya {umur} tahun.")
```
