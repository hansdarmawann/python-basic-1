## 1) Dasar String


Di sel ini kita melihat bahwa string adalah kumpulan karakter yang berurutan. Karena berupa urutan, kita bisa menghitung panjangnya dengan `len()` dan mengambil karakter tertentu lewat indeks. Ini berguna saat kamu perlu memproses teks per huruf. Pemahaman ini jadi dasar untuk manipulasi string yang lebih lanjut.


```python
# Sel ini menunjukkan bahwa string adalah teks dan punya panjang serta indeks karakter.

teks = "Belajar Python"
print(teks)
print("Panjang:", len(teks))
print("Karakter pertama:", teks[0])
```

## 2) Slicing


Bagian ini membahas slicing, yaitu mengambil potongan string. Pola `[awal:akhir]` sangat membantu kalau kita hanya butuh sebagian teks, misalnya nama depan atau kode tertentu. Perhatikan bahwa batas akhir tidak ikut terambil, ini detail kecil yang penting. Setelah terbiasa, slicing akan terasa sangat praktis.


```python
# Sel ini menampilkan teknik slicing untuk mengambil sebagian teks.

teks = "Belajar Python"
print(teks[0:7])   # Belajar
print(teks[8:14])  # Python
print(teks[-6:])   # Python
```

## 3) String Tidak Bisa Diubah Langsung


Di sini kita belajar konsep immutable pada string. Artinya, karakter string tidak bisa diubah langsung pada posisi tertentu. Solusinya adalah membuat string baru dengan menggabungkan bagian yang dibutuhkan. Ini terlihat sepele, tapi sangat penting supaya kamu paham perilaku data di Python.


```python
# Sel ini menjelaskan string immutable, lalu solusi membuat string baru.

nama = "andi"
# nama[0] = "A"  # Ini error
nama_baru = "A" + nama[1:]
print(nama_baru)
```

## 4) Method String


Sel ini memperkenalkan beberapa method string yang sering dipakai sehari-hari. `strip()` membersihkan spasi berlebih, `upper()` mengubah jadi huruf besar, dan `replace()` mengganti teks tertentu. Ada juga operator `in` untuk mengecek apakah kata tertentu ada di dalam kalimat. Kombinasi ini cukup kuat untuk banyak kebutuhan olah teks dasar.


```python
# Sel ini mencoba beberapa method string yang sangat sering dipakai.

kalimat = "  saya suka python  "
print(kalimat.strip())
print(kalimat.upper())
print(kalimat.replace("python", "data"))
print("python" in kalimat)
```

## 5) f-string


Di bagian akhir, kita pakai f-string untuk format teks yang lebih enak dibaca. Dibanding cara lama dengan banyak tanda plus, f-string lebih rapi dan minim kesalahan. Kamu tinggal sisipkan variabel di dalam `{}`. Saat menampilkan laporan atau pesan ke pengguna, cara ini sangat direkomendasikan.


```python
# Sel ini mencontohkan f-string untuk format teks yang mudah dibaca.

nama = "Rani"
umur = 22
print(f"Nama saya {nama}, umur saya {umur} tahun.")
```
