## 5.2. Dasar Struktur: Cara Mendefinisikan String
String adalah struktur data untuk teks. Kita bisa mendefinisikannya dengan petik tunggal, petik ganda, triple quote, atau lewat `str()`. Di praktik nyata, kamu bebas pilih cara yang paling nyaman selama konsisten. Untuk teks multi-baris, triple quote biasanya paling enak dipakai.


```python
# Beberapa cara mendefinisikan string
s1 = 'Halo'
s2 = "Python"
s3 = """Baris 1
Baris 2"""
s4 = str(12345)

print(s1, type(s1))
print(s2, type(s2))
print(s3)
print(s4, type(s4))
```

## 5.3. Sifat String (`immutable`, `indexable`, `duplicable`)
String itu `immutable`, artinya karakter di dalamnya tidak bisa diubah langsung. String juga `indexable`, jadi bisa diakses lewat index seperti `teks[0]` atau `teks[-1]`. Selain itu string `duplicable`, artinya bisa diulang dengan operator `*`. Tiga sifat ini wajib dipahami karena sering muncul di error pemula.


```python
# Sifat penting string
teks = "koding"
print("Indexable:", teks[0], teks[-1])
print("Duplicable:", teks * 2)

# Immutable: baris di bawah ini akan error jika dibuka
# teks[0] = "K"
```

## 5.4. Cara Akses String (Index dan Slicing)
Akses string tidak cuma satu pola. Kamu bisa ambil satu karakter, potongan dari kiri ke kanan, potongan dari posisi tertentu sampai akhir, atau pakai langkah (`step`) untuk loncat karakter. Kamu juga bisa membalik string dengan slicing. Ini sangat berguna saat parsing dan pembersihan teks.


```python
# Variasi akses string
teks = "BelajarPython"
print(teks[0])        # karakter pertama
print(teks[-1])       # karakter terakhir
print(teks[0:7])      # dari index 0 s.d 6
print(teks[7:])       # dari index 7 sampai akhir
print(teks[:7])       # dari awal sampai index 6
print(teks[::2])      # loncat 2 karakter
print(teks[::-1])     # dibalik
```

## 5.5. Operasi String yang Umum
Operasi dasar string meliputi penggabungan (`+`), pengulangan (`*`), cek keberadaan (`in`), perbandingan, dan menghitung panjang (`len`). Operasi-operasi ini sangat sering dipakai di validasi input dan formatting output. Kalau kamu kuasai bagian ini, kemampuan olah teksmu naik cepat. Biasakan juga membaca hasil operasi agar tidak tertukar antara string dan angka.


```python
# Operasi string yang umum
nama_depan = "Budi"
nama_belakang = "Santoso"

print(nama_depan + " " + nama_belakang)     # gabung
print("ha" * 3)                              # ulang
print("thon" in "Python")                   # cek isi
print("apel" < "jeruk")                     # perbandingan
print(len("Belajar"))                        # panjang string
```

## 5.6. Method String Penting + Penjelasan
Di bawah ini adalah method string yang paling sering dipakai:
- `strip()` menghapus spasi di kiri/kanan.
- `lower()` dan `upper()` mengubah kapitalisasi huruf.
- `replace(a, b)` mengganti teks `a` menjadi `b`.
- `split()` memecah string menjadi list.
- `join(list)` menggabungkan list menjadi string.
- `startswith()` dan `endswith()` untuk cek awalan/akhiran.


```python
# Method string penting
kalimat = "  Saya Suka Python  "

print(kalimat.strip())
print(kalimat.lower())
print(kalimat.upper())
print(kalimat.replace("Python", "Data"))
print(kalimat.strip().split())
print("-".join(["python", "basic", "id"]))
print(kalimat.startswith("  Sa"))
print(kalimat.endswith("  "))
```
