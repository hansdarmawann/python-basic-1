## Tujuan Belajar
- Memahami string sebagai urutan karakter.
- Mengakses karakter dengan indeks dan slicing.
- Menggunakan method string penting.
- Memformat teks dengan f-string.


## 1) Dasar String


**Penjelasan Code 1:** Sel ini menunjukkan bahwa string adalah teks dan punya panjang serta indeks karakter.


```python
# Sel ini menunjukkan bahwa string adalah teks dan punya panjang serta indeks karakter.

teks = "Belajar Python"
print(teks)
print("Panjang:", len(teks))
print("Karakter pertama:", teks[0])
```

## 2) Slicing


**Penjelasan Code 2:** Sel ini menampilkan teknik slicing untuk mengambil sebagian teks.


```python
# Sel ini menampilkan teknik slicing untuk mengambil sebagian teks.

teks = "Belajar Python"
print(teks[0:7])   # Belajar
print(teks[8:14])  # Python
print(teks[-6:])   # Python
```

## 3) String Tidak Bisa Diubah Langsung


**Penjelasan Code 3:** Sel ini menjelaskan string immutable, lalu solusi membuat string baru.


```python
# Sel ini menjelaskan string immutable, lalu solusi membuat string baru.

nama = "andi"
# nama[0] = "A"  # Ini error
nama_baru = "A" + nama[1:]
print(nama_baru)
```

## 4) Method String


**Penjelasan Code 4:** Sel ini mencoba beberapa method string yang sangat sering dipakai.


```python
# Sel ini mencoba beberapa method string yang sangat sering dipakai.

kalimat = "  saya suka python  "
print(kalimat.strip())
print(kalimat.upper())
print(kalimat.replace("python", "data"))
print("python" in kalimat)
```

## 5) f-string


**Penjelasan Code 5:** Sel ini mencontohkan f-string untuk format teks yang mudah dibaca.


```python
# Sel ini mencontohkan f-string untuk format teks yang mudah dibaca.

nama = "Rani"
umur = 22
print(f"Nama saya {nama}, umur saya {umur} tahun.")
```
