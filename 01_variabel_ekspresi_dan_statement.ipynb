## 1) Nilai dan Tipe Data


Di sel ini kita mulai dari fondasi paling dasar: mengenali tipe data. Saat belajar Python, penting sekali memahami bahwa angka bulat, angka desimal, teks, dan nilai benar/salah diperlakukan berbeda. Fungsi `type()` membantu kita melihat tipe dari sebuah nilai dengan jelas. Dengan kebiasaan ini, kamu akan lebih mudah menghindari bug saat program mulai panjang.


```python
# Menampilkan tipe data dari beberapa nilai.

print(type(10))
print(type(3.14))
print(type("Halo"))
print(type(True))
```

## 2) Variabel


Sekarang kita masuk ke konsep variabel, yaitu tempat menyimpan data dengan nama yang mudah diingat. Anggap saja variabel seperti label pada kotak penyimpanan: isi kotaknya bisa kita panggil lagi kapan pun dibutuhkan. Di contoh ini, kita menyimpan nama dan umur lalu menampilkannya ke layar. Ini pola yang nanti akan sering banget dipakai di program nyata.


```python
# Sel ini menunjukkan cara menyimpan data ke variabel, lalu menampilkannya kembali.

nama = "Budi"
umur = 25
print("Nama:", nama)
print("Umur:", umur)
```

## 3) Operator Dasar


Bagian ini memperkenalkan operator matematika dasar yang paling sering kamu pakai. Kita lihat hasil penjumlahan, pengurangan, perkalian, pembagian biasa, pembagian bulat, sisa bagi, sampai pangkat. Coba perhatikan perbedaan `/` dan `//` karena ini sering bikin pemula tertukar. Kalau kamu sudah nyaman di sini, menghitung data di Python akan terasa jauh lebih mudah.


```python
# Sel ini mendemokan operator aritmatika yang paling sering dipakai di program dasar.

a = 10
b = 3
print("Tambah:", a + b)
print("Kurang:", a - b)
print("Kali:", a * b)
print("Bagi:", a / b)
print("Bagi bulat:", a // b)
print("Sisa bagi:", a % b)
print("Pangkat:", a ** b)
```

## 4) Input Pengguna


Di sini program mulai berinteraksi dengan pengguna lewat `input()`. Saat pengguna mengetik, Python akan menerima nilainya sebagai teks. Karena itu contoh ini simpel dulu: kita tampilkan kembali inputnya sebagai sapaan. Tujuannya agar kamu paham alur masuknya data dari luar ke program.


```python
# Input dari pengguna selalu string, jadi dikonversi ke int.

nama = input("Siapa nama kamu? ")
print("Halo", nama)
```

## 5) Konversi Tipe


Nah, sel ini penting karena menunjukkan proses konversi tipe data. Hasil `input()` awalnya berupa string, jadi harus diubah ke `int` kalau mau dihitung sebagai angka. Setelah dikonversi, barulah operasi matematika berjalan sesuai harapan. Pola ini wajib kamu ingat saat membuat kalkulator atau form input angka.


```python
# Input dari pengguna selalu string, jadi dikonversi ke int.

angka = input("Masukkan angka: ")
angka_int = int(angka)
print("Hasil kali 2:", angka_int * 2)
```
