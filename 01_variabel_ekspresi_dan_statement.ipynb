## 1) Nilai dan Tipe Data


**Penjelasan Code 1:** Sel ini menampilkan tipe data dasar Python agar kita tahu bedanya angka bulat, desimal, teks, dan boolean.


```python
# Menampilkan tipe data dari beberapa nilai.

print(type(10))
print(type(3.14))
print(type("Halo"))
print(type(True))
```

## 2) Variabel


**Penjelasan Code 2:** Sel ini menunjukkan cara menyimpan data ke variabel, lalu menampilkannya kembali.


```python
# Sel ini menunjukkan cara menyimpan data ke variabel, lalu menampilkannya kembali.

nama = "Budi"
umur = 25
print("Nama:", nama)
print("Umur:", umur)
```

## 3) Operator Dasar


**Penjelasan Code 3:** Sel ini mendemokan operator aritmatika yang paling sering dipakai di program dasar.


```python
# Sel ini mendemokan operator aritmatika yang paling sering dipakai di program dasar.

a = 10
b = 3
print("Tambah:", a + b)
print("Kurang:", a - b)
print("Kali:", a * b)
print("Bagi:", a / b)
print("Bagi bulat:", a // b)
print("Sisa bagi:", a % b)
print("Pangkat:", a ** b)
```

## 4) Input Pengguna


**Penjelasan Code 4:** Sel ini memakai `input()` untuk menerima data teks dari pengguna.


```python
# Input dari pengguna selalu string, jadi dikonversi ke int.

nama = input("Siapa nama kamu? ")
print("Halo", nama)
```

## 5) Konversi Tipe


**Penjelasan Code 5:** Sel ini mengubah input teks menjadi angka dengan `int()` sebelum dihitung.


```python
# Input dari pengguna selalu string, jadi dikonversi ke int.

angka = input("Masukkan angka: ")
angka_int = int(angka)
print("Hasil kali 2:", angka_int * 2)
```
