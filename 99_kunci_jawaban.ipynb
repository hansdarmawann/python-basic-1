## Bab 1 - Variabel, Ekspresi, dan Statement
### Soal 1: Konversi Celcius ke Fahrenheit


```python
celcius = float(input('Masukkan suhu Celcius: '))
fahrenheit = (celcius * 9/5) + 32
print('Fahrenheit:', fahrenheit)
```

### Soal 2: Luas Persegi Panjang


```python
panjang = float(input('Panjang: '))
lebar = float(input('Lebar: '))
luas = panjang * lebar
print('Luas:', luas)
```

## Bab 2 - Percabangan
### Soal 1: Cek Genap/Ganjil


```python
n = int(input('Masukkan bilangan: '))
if n % 2 == 0:
    print('Genap')
else:
    print('Ganjil')
```

### Soal 2: Kategori Umur


```python
umur = int(input('Masukkan umur: '))
if umur < 13:
    print('Anak')
elif umur < 18:
    print('Remaja')
else:
    print('Dewasa')
```

## Bab 3 - Fungsi
### Soal 1: `hitung_diskon(harga, persen_diskon)`


```python
def hitung_diskon(harga, persen_diskon):
    return harga - (harga * persen_diskon / 100)

print(hitung_diskon(100000, 20))
```

### Soal 2: `cek_lulus(nilai)`


```python
def cek_lulus(nilai):
    return nilai >= 75

print(cek_lulus(80))
print(cek_lulus(60))
```

## Bab 4 - Perulangan
### Soal 1: Angka 1-20 yang habis dibagi 3


```python
for i in range(1, 21):
    if i % 3 == 0:
        print(i)
```

### Soal 2: Rata-rata daftar angka


```python
angka = [10, 20, 30, 40, 50]
rata_rata = sum(angka) / len(angka)
print('Rata-rata:', rata_rata)
```

## Bab 5 - String
### Soal 1: Ambil domain email


```python
email = input('Masukkan email: ').strip()
domain = email.split('@')[-1]
print('Domain:', domain)
```

### Soal 2: Cek palindrome (abaikan kapital)


```python
teks = input('Masukkan teks: ').strip().lower()
if teks == teks[::-1]:
    print('Palindrome')
else:
    print('Bukan palindrome')
```

### Soal 3: 5 karakter awal dan akhir


```python
teks = input('Masukkan teks: ')
print('5 awal :', teks[:5])
print('5 akhir:', teks[-5:])
```

## Bab 6 - List
### Soal 1: Tampilkan angka genap dari 1-10


```python
angka = list(range(1, 11))
genap = [x for x in angka if x % 2 == 0]
print(genap)
```

### Soal 2: Urutkan nama lalu tampilkan dari belakang


```python
nama = ['Rani', 'Budi', 'Andi', 'Caca']
nama.sort()
print(nama[::-1])
```

### Soal 3: Bandingkan `remove()` dan `pop()`


```python
data = ['a', 'b', 'c', 'b']

data_remove = data.copy()
data_remove.remove('b')  # hapus nilai pertama yang cocok
print('Setelah remove:', data_remove)

data_pop = data.copy()
hasil_pop = data_pop.pop(1)  # hapus berdasarkan index
print('Setelah pop   :', data_pop)
print('Nilai pop     :', hasil_pop)
```

## Bab 7 - Dictionary
### Soal 1: Biodata sederhana


```python
biodata = {'nama': 'Dina', 'umur': 22, 'kota': 'Bandung'}
print(biodata)
```

### Soal 2: Frekuensi huruf kata `banana`


```python
kata = 'banana'
freq = {}
for huruf in kata:
    freq[huruf] = freq.get(huruf, 0) + 1
print(freq)
```

### Soal 3: Bedakan akses `[]` vs `.get()`


```python
data = {'nama': 'Ari'}

# print(data['umur'])  # KeyError jika key tidak ada
print(data.get('umur'))         # None
print(data.get('umur', 'N/A'))  # nilai default
```

## Bab 8 - Set
### Soal 1: Ambil nama unik dari list berulang


```python
nama = ['Ana', 'Budi', 'Ana', 'Caca', 'Budi']
unik = set(nama)
print(unik)
```

### Soal 2: Hobi yang sama dari dua set


```python
hobi_a = {'membaca', 'lari', 'musik'}
hobi_b = {'musik', 'memasak', 'lari'}
print(hobi_a.intersection(hobi_b))
```

### Soal 3: Bedakan `remove()` dan `discard()`


```python
s = {'apel', 'jeruk'}

# s.remove('mangga')  # KeyError karena item tidak ada
s.discard('mangga')   # aman, tidak error
print(s)
```

## Bab 9 - Tuple
### Soal 1: Tuple 5 angka, tampilkan terbesar


```python
t = (12, 7, 30, 4, 18)
print(max(t))
```

### Soal 2: Unpack tuple `(nama, umur, kota)`


```python
data = ('Nina', 24, 'Surabaya')
nama, umur, kota = data
print(nama, umur, kota)
```

### Soal 3: Kapan pakai list vs tuple?
- Pakai `list` kalau datanya sering berubah (tambah/hapus/update).
- Pakai `tuple` kalau datanya tetap dan ingin lebih aman dari perubahan.


## Bab 10 - List vs Dict vs Set vs Tuple
### Soal 1: Nilai unik dan urut naik


```python
nilai = [80, 90, 80, 70, 90]
unik_urut = sorted(set(nilai))
print('Nilai unik urut:', unik_urut)
```

### Soal 2: Dict biodata dan kalimat perkenalan


```python
biodata = {'nama': 'Salsa', 'kota': 'Yogyakarta', 'hobi': 'membaca'}
print(f"Halo, saya {biodata['nama']} dari {biodata['kota']}, hobi saya {biodata['hobi']}.")
```

### Soal 3: Tuple koordinat rumah dan kantor


```python
rumah = (2, 5)
kantor = (10, 14)
selisih_x = kantor[0] - rumah[0]
selisih_y = kantor[1] - rumah[1]
print('Selisih x:', selisih_x)
print('Selisih y:', selisih_y)
```
