## 1) Nilai dan Tipe Data


```python
print(type(10))
print(type(3.14))
print(type("Halo"))
print(type(True))
```

## 2) Variabel


```python
nama = "Budi"
umur = 25
print("Nama:", nama)
print("Umur:", umur)
```

## 3) Operator Dasar


```python
a = 10
b = 3
print("Tambah:", a + b)
print("Kurang:", a - b)
print("Kali:", a * b)
print("Bagi:", a / b)
print("Bagi bulat:", a // b)
print("Sisa bagi:", a % b)
print("Pangkat:", a ** b)
```

## 4) Input Pengguna


```python
nama = input("Siapa nama kamu? ")
print("Halo", nama)
```

## 5) Konversi Tipe


```python
angka = input("Masukkan angka: ")
angka_int = int(angka)
print("Hasil kali 2:", angka_int * 2)
```
