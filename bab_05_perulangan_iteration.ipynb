# Bab 5: Perulangan (Iteration)


## Tujuan Belajar
- Memahami `while` dan `for`.
- Menggunakan `break` dan `continue`.
- Mengenal pola umum perulangan (hitung, jumlah, max/min).


## 1) while


```python
angka = 1
while angka <= 5:
    print("Angka:", angka)
    angka += 1
```

## 2) for


```python
for nama in ["Ani", "Budi", "Cici"]:
    print("Halo", nama)
```

## 3) break dan continue


```python
for i in range(1, 8):
    if i == 6:
        break
    if i % 2 == 0:
        continue
    print(i)
```

## 4) Pola Menjumlahkan


```python
total = 0
for n in [10, 20, 30]:
    total += n
print("Total:", total)
```

## 5) Pola Cari Nilai Maksimum


```python
data = [8, 3, 15, 6, 20, 1]
terbesar = data[0]
for n in data:
    if n > terbesar:
        terbesar = n
print("Nilai terbesar:", terbesar)
```

## Latihan
- Cetak angka 1-20 yang habis dibagi 3.
- Hitung rata-rata dari daftar angka.
