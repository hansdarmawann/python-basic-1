## 1) Boolean dan Perbandingan


Di sel ini kita belajar bagaimana komputer mengambil keputusan dengan nilai boolean. Operasi perbandingan seperti `>=`, `==`, dan `!=` akan menghasilkan `True` atau `False`. Hasil inilah yang nanti dipakai oleh `if` untuk menentukan jalur program. Kalau konsep ini sudah klik, percabangan akan terasa logis dan tidak menakutkan.


```python
# Sel ini menunjukkan contoh hasil perbandingan yang menghasilkan nilai `True` atau `False`.

usia = 18
print(usia >= 17)
print(usia == 20)
print(usia != 10)
```

## 2) if - elif - else


Sekarang kita praktik percabangan `if`, `elif`, dan `else` untuk menentukan grade. Program akan mengecek kondisi dari atas ke bawah sampai menemukan yang cocok. Jadi urutan kondisi itu penting, jangan sampai aturan yang umum ditaruh sebelum aturan yang lebih spesifik. Dengan pola ini, kamu bisa membuat sistem keputusan sederhana dengan rapi.


```python
# Menentukan alur program berdasarkan beberapa kondisi.

nilai = 78

if nilai >= 85:
    grade = "A"
elif nilai >= 70:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)
```

## 3) Operator Logika


Sel ini memperkenalkan operator logika `and`. Artinya, kedua syarat harus benar supaya hasil akhirnya benar. Contoh seperti ini sering dipakai saat validasi data, misalnya syarat umur dan dokumen harus sama-sama terpenuhi. Coba ganti nilainya dan lihat bagaimana hasilnya berubah.


```python
# Sel ini memperlihatkan operator logika `and` untuk menggabungkan dua syarat.

umur = 20
punya_ktp = True

if umur >= 17 and punya_ktp:
    print("Boleh daftar")
else:
    print("Belum boleh daftar")
```

## 4) Menangani Error Input


Bagian ini mengajarkan cara bikin program lebih tahan banting dengan `try-except`. Kalau pengguna memasukkan data yang tidak sesuai, program tidak langsung crash. Sebaliknya, kita bisa menampilkan pesan yang ramah dan mudah dipahami. Ini kebiasaan bagus sejak awal supaya pengalaman pengguna tetap nyaman.


```python
# Input dari pengguna selalu string, jadi dikonversi ke int.

try:
    angka = int(input("Masukkan angka bulat: "))
    print("Kuadrat:", angka ** 2)
except:
    print("Input tidak valid. Harus angka bulat.")
```
