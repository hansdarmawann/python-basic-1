# Bab 3: Percabangan (Conditional Execution)


## Tujuan Belajar
- Memahami nilai boolean (`True`/`False`).
- Menggunakan `if`, `elif`, `else`.
- Menggabungkan syarat dengan `and`, `or`, `not`.
- Menangani input yang salah dengan `try`/`except`.


## 1) Boolean dan Perbandingan


```python
usia = 18
print(usia >= 17)
print(usia == 20)
print(usia != 10)
```

## 2) if - elif - else


```python
nilai = 78

if nilai >= 85:
    grade = "A"
elif nilai >= 70:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)
```

## 3) Operator Logika


```python
umur = 20
punya_ktp = True

if umur >= 17 and punya_ktp:
    print("Boleh daftar")
else:
    print("Belum boleh daftar")
```

## 4) Menangani Error Input


```python
try:
    angka = int(input("Masukkan angka bulat: "))
    print("Kuadrat:", angka ** 2)
except:
    print("Input tidak valid. Harus angka bulat.")
```
