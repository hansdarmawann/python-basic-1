## 6.2. Dasar Struktur: Cara Mendefinisikan List
List menyimpan kumpulan data berurutan. Kita bisa membuat list dengan literal `[]` atau dengan konstruktor `list()`. `list()` juga berguna untuk mengubah data lain jadi list, misalnya dari string. Pahami ini dulu karena list adalah struktur data paling sering dipakai pemula.


```python
# Cara mendefinisikan list
l1 = [10, 20, 30]
l2 = list(("apel", "jeruk", "mangga"))
l3 = list("ABC")

print(l1)
print(l2)
print(l3)
```

## 6.3. Sifat List (`mutable`, `indexable`, `duplicable`)
List itu `mutable`, jadi elemennya bisa diubah, ditambah, atau dihapus. List juga `indexable`, artinya bisa diakses pakai index positif/negatif. Selain itu list `duplicable`, jadi bisa diulang dengan operator `*`. Karena fleksibel, list sering jadi pilihan awal saat menyimpan kumpulan data.


```python
# Sifat list
data = ["a", "b", "c"]
print("Indexable:", data[0], data[-1])

data[1] = "B"    # mutable
print("Mutable:", data)

print("Duplicable:", data * 2)
```

## 6.4. Cara Akses List (Index dan Slicing)
Akses list bisa dilakukan dengan banyak pola. Kamu bisa ambil item tunggal, sebagian range, dari awal ke batas tertentu, dari batas tertentu ke akhir, pakai step, sampai membalik list. Slicing ini bukan cuma untuk lihat data, tapi sering dipakai saat preprocessing data. Biasakan baca hasil tiap pola agar cepat paham.


```python
# Variasi akses list
angka = [0, 1, 2, 3, 4, 5, 6]
print(angka[2])
print(angka[-1])
print(angka[1:5])
print(angka[:4])
print(angka[3:])
print(angka[::2])
print(angka[::-1])
```

## 6.5. Method List Penting + Penjelasan
Method list yang paling sering dipakai:
- `append(x)` menambah item di akhir list.
- `extend(iterable)` menambah banyak item sekaligus.
- `insert(i, x)` menyisipkan item di posisi tertentu.
- `remove(x)` menghapus item berdasarkan nilai.
- `pop()` menghapus item terakhir dan mengembalikannya.
- `sort()` mengurutkan list.
- `reverse()` membalik urutan list.


```python
# Method list penting
buah = ["apel", "jeruk"]
buah.append("mangga")
buah.extend(["pisang", "kiwi"])
buah.insert(1, "anggur")
buah.remove("jeruk")
item_akhir = buah.pop()
buah.sort()
buah.reverse()

print("List akhir:", buah)
print("Item pop:", item_akhir)
```

## 6.6. Operasi List yang Umum
Selain method, list punya operasi penting seperti gabung list dengan `+`, cek keberadaan item dengan `in`, dan menghitung panjang dengan `len()`. Di dunia nyata, operasi ini muncul saat validasi data, filtering sederhana, dan proses agregasi. Gabungkan operasi dan method agar alur kerjamu efisien. Kuncinya: coba, lihat output, lalu ulangi.


```python
# Operasi list yang umum
a = [1, 2, 3]
b = [4, 5]
print(a + b)
print(2 in a)
print(len(a))
```
