## 1) Fungsi Bawaan


**Penjelasan Code 1:** Sel ini memakai fungsi bawaan seperti `len()`, `max()`, dan `min()` pada list angka.


```python
# Sel ini memakai fungsi bawaan seperti `len()`, `max()`, dan `min()` pada list angka.

angka = [10, 2, 30, 4]
print("Panjang:", len(angka))
print("Nilai terbesar:", max(angka))
print("Nilai terkecil:", min(angka))
```

## 2) Membuat Fungsi Sendiri


**Penjelasan Code 2:** Sel ini membuat fungsi sederhana `sapa()` dengan satu parameter.


```python
# Mendefinisikan fungsi agar kode bisa dipakai ulang.

def sapa(nama):
    print(f"Halo, {nama}! Selamat belajar Python.")

sapa("Rina")
```

## 3) Parameter dan Return


**Penjelasan Code 3:** Sel ini membuat fungsi yang mengembalikan nilai lewat `return`.


```python
# Mendefinisikan fungsi agar kode bisa dipakai ulang.

def luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

hasil = luas_persegi_panjang(8, 5)
print("Luas:", hasil)
```

## 4) Fungsi untuk Reuse


**Penjelasan Code 4:** Sel ini menunjukkan manfaat fungsi untuk dipakai berulang di beberapa data.


```python
# Mendefinisikan fungsi agar kode bisa dipakai ulang.

def konversi_c_ke_f(c):
    return (c * 9/5) + 32

for c in [0, 25, 30]:
    print(c, "C =", konversi_c_ke_f(c), "F")
```
