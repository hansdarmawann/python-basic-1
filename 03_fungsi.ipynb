## 3.2. Fungsi Bawaan


Di awal bab fungsi, kita kenalan dulu dengan fungsi bawaan Python. Contoh `len()`, `max()`, dan `min()` menunjukkan bahwa banyak pekerjaan umum sudah disediakan oleh Python. Jadi kita tidak selalu harus menulis semuanya dari nol. Ini membuat proses coding jadi lebih cepat dan lebih fokus ke logika utama.


```python
# Sel ini memakai fungsi bawaan seperti `len()`, `max()`, dan `min()` pada list angka.

angka = [10, 2, 30, 4]
print("Panjang:", len(angka))
print("Nilai terbesar:", max(angka))
print("Nilai terkecil:", min(angka))
```

## 3.3. Membuat Fungsi Sendiri


Sekarang kita membuat fungsi sendiri dengan `def`. Fungsi membantu kita menyimpan langkah yang sering dipakai agar tidak menulis kode berulang-ulang. Di contoh ini, fungsi menerima parameter `nama` lalu membentuk sapaan otomatis. Kalau nanti kamu butuh sapaan ke banyak orang, tinggal panggil fungsi yang sama.


```python
# Mendefinisikan fungsi agar kode bisa dipakai ulang.

def sapa(nama):
    print(f"Halo, {nama}! Selamat belajar Python.")

sapa("Rina")
```

## 3.4. Parameter dan Return


Sel ini menekankan peran `return`, yaitu mengembalikan hasil dari fungsi. Bedanya dengan `print`, nilai yang di-`return` bisa disimpan lagi ke variabel dan dipakai di proses berikutnya. Ini penting untuk membuat fungsi yang benar-benar berguna di alur program. Biasakan memikirkan: fungsi ini menghasilkan nilai apa?


```python
# Mendefinisikan fungsi agar kode bisa dipakai ulang.

def luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

hasil = luas_persegi_panjang(8, 5)
print("Luas:", hasil)
```

## 3.5. Fungsi untuk Reuse


Di sini kamu lihat manfaat fungsi dalam skenario berulang. Kita bikin satu rumus konversi suhu, lalu memakainya berkali-kali untuk nilai yang berbeda. Pola seperti ini membuat kode lebih bersih, mudah dirawat, dan minim salah hitung. Semakin sering kamu memakai fungsi, semakin terstruktur cara berpikirmu saat coding.


```python
# Mendefinisikan fungsi agar kode bisa dipakai ulang.

def konversi_c_ke_f(c):
    return (c * 9/5) + 32

for c in [0, 25, 30]:
    print(c, "C =", konversi_c_ke_f(c), "F")
```
