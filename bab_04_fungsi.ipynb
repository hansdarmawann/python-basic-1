# Bab 4: Fungsi


## Tujuan Belajar
- Memahami kenapa fungsi penting.
- Membuat fungsi dengan `def`.
- Menggunakan parameter dan nilai balik (`return`).
- Memakai fungsi bawaan Python.


## 1) Fungsi Bawaan


```python
angka = [10, 2, 30, 4]
print("Panjang:", len(angka))
print("Nilai terbesar:", max(angka))
print("Nilai terkecil:", min(angka))
```

## 2) Membuat Fungsi Sendiri


```python
def sapa(nama):
    print(f"Halo, {nama}! Selamat belajar Python.")

sapa("Rina")
```

## 3) Parameter dan Return


```python
def luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

hasil = luas_persegi_panjang(8, 5)
print("Luas:", hasil)
```

## 4) Fungsi untuk Reuse


```python
def konversi_c_ke_f(c):
    return (c * 9/5) + 32

for c in [0, 25, 30]:
    print(c, "C =", konversi_c_ke_f(c), "F")
```
